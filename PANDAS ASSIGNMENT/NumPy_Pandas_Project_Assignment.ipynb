## Question 1

Create a NumPy array using the salary values:

Input:
```
[50000,42000,65000,58000,45000]
```

Output:
```
array([50000,42000,65000,58000,45000])
```

## Question 2

Using the NumPy array, find Total Salary, Average Salary and Maximum Salary.

Output:
```
Total Salary = 260000
Average Salary = 52000.0
Maximum Salary = 65000
```

````python
import numpy as np

salary_array = np.array([50000, 42000, 65000, 58000, 45000])

total_salary = np.sum(salary_array)
average_salary = np.mean(salary_array)
maximum_salary = np.max(salary_array)

print(f"Total Salary = {total_salary}")
print(f"Average Salary = {average_salary}")
print(f"Maximum Salary = {maximum_salary}")
````

## Question 3

Convert the NumPy array into a Pandas DataFrame.

Column Name: **Salary**

````python
import numpy as np
import pandas as pd

salary_array = np.array([50000, 42000, 65000, 58000, 45000])
df_salary = pd.DataFrame(salary_array, columns=['Salary'])
print(df_salary)
````

## Question 4

Create the Employee DataFrame using the given dataset.

## Question 13

Find the **Average Salary** of each Department using **groupby()**.

Expected Output:
Finance -> 58000
HR -> 43500
IT -> 57500

````python
# Assuming 'df' is already created from Question 4
avg_salary = df.groupby('Department')['Salary'].mean()
for dept, avg in avg_salary.items():
    print(f"{dept} -> {int(avg)}")
````

## Question 15

Create another DataFrame containing **EmpID** and **Experience** and merge it with the Employee DataFrame.

Second DataFrame:

| EmpID | Experience |
|------:|-----------:|
|101|2|
|102|5|
|103|3|
|104|6|
|105|4|

Expected Output: Final merged DataFrame with Experience column.

````python
import pandas as pd

# Assuming original 'df' exists from Question 4
experience_data = {
    'EmpID': [101, 102, 103, 104, 105],
    'Experience': [2, 5, 3, 6, 4]
}

df_exp = pd.DataFrame(experience_data)
final_df = pd.merge(df, df_exp, on='EmpID')
print(final_df)
````
