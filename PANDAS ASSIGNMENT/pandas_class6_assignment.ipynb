# Q1. Display Unique Values

## Problem Statement
Create a DataFrame with a City column and display all unique cities.

## Sample Input
```python
City = ["Bhopal", "Indore", "Bhopal", "Delhi"]
```

## Sample Output
```python
['Bhopal' 'Indore' 'Delhi']
```


````python
import pandas as pd

df = pd.DataFrame({'City': ["Bhopal", "Indore", "Bhopal", "Delhi"]})
print(df['City'].unique())
````


# Q4. Group By City

## Problem Statement
Find average marks city-wise using groupby().

## Sample Input
```python
City   Marks
Bhopal 80
Bhopal 90
Delhi  70
```

## Sample Output
```python
Bhopal 85
Delhi  70
```


````python
import pandas as pd

df = pd.DataFrame({
    'City': ["Bhopal", "Bhopal", "Delhi"],
    'Marks': [80, 90, 70]
})
print(df.groupby('City')['Marks'].mean())
````


# Q5. Sort Values

## Problem Statement
Sort students by Marks in ascending order.

## Sample Output
```python
Sorted DataFrame
```


````python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Aman', 'Priya', 'Rahul'],
    'Marks': [85, 92, 78]
})
sorted_df = df.sort_values(by='Marks', ascending=True)
print("Sorted DataFrame")
print(sorted_df)
````


# Q6. Sort Values Descending

## Problem Statement
Sort students by Marks in descending order.

## Sample Output
```python
Sorted DataFrame
```


````python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Aman', 'Priya', 'Rahul'],
    'Marks': [85, 92, 78]
})
sorted_df = df.sort_values(by='Marks', ascending=False)
print("Sorted DataFrame")
print(sorted_df)
````


# Q7. Filter Records

## Problem Statement
Display students having marks greater than 75.

## Sample Output
```python
Filtered rows
```


````python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Aman', 'Priya', 'Rahul', 'Neha'],
    'Marks': [85, 60, 78, 90]
})
filtered_df = df[df['Marks'] > 75]
print("Filtered rows")
print(filtered_df)
````


# Q8. Multiple Conditions

## Problem Statement
Display students whose marks are greater than 70 and age is greater than 20.

## Sample Output
```python
Filtered rows
```


````python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Aman', 'Priya', 'Rahul', 'Neha'],
    'Marks': [75, 85, 65, 90],
    'Age': [19, 22, 21, 20]
})
filtered_df = df[(df['Marks'] > 70) & (df['Age'] > 20)]
print("Filtered rows")
print(filtered_df)
````


# Q9. Group By Department

## Problem Statement
Create Employee and Department data and find total salary department-wise.

## Sample Output
```python
HR       50000
IT      120000
```


````python
import pandas as pd

df = pd.DataFrame({
    'Department': ['HR', 'IT', 'HR', 'IT'],
    'Salary': [25000, 60000, 25000, 60000]
})
print(df.groupby('Department')['Salary'].sum())
````


# Q10. Complete Practice Question

## Problem Statement
Create a DataFrame containing:
- Name
- Age
- City
- Marks

Perform:
1. Show unique cities
2. Count city frequency
3. Filter marks > 80
4. Sort by marks

## Sample Output
```python
Required Results
```


````python
import pandas as pd

# Creating the DataFrame
df = pd.DataFrame({
    'Name': ['Amit', 'Pooja', 'Rohan', 'Sneha', 'Vikram'],
    'Age': [21, 23, 19, 22, 20],
    'City': ['Bhopal', 'Indore', 'Bhopal', 'Delhi', 'Indore'],
    'Marks': [78, 85, 92, 74, 88]
})

print("--- 1. Unique Cities ---")
print(df['City'].unique())

print("\n--- 2. City Frequency ---")
print(df['City'].value_counts())

print("\n--- 3. Filter Marks > 80 ---")
print(df[df['Marks'] > 80])

print("\n--- 4. Sort by Marks ---")
print(df.sort_values(by='Marks', ascending=False))
````
