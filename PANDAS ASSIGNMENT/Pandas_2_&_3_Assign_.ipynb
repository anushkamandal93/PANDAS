# Q1. Display First Rows

## Problem Statement
Create the following DataFrame:

| Name | Age |
|------|-----|
| Aman | 21 |
| Rahul | 22 |
| Neha | 20 |
| Priya | 23 |
| Ravi | 24 |

Display the first 3 rows using `head()`.

## Sample Output
```python
First 3 rows of DataFrame
```


````python
import pandas as pd

# 1. Create the DataFrame
data = {
    "Name": ["Aman", "Rahul", "Neha", "Priya", "Ravi"],
    "Age": [21, 22, 20, 23, 24]
}
df = pd.DataFrame(data)

# 2. Display the first 3 rows using head()
first_three_rows = df.head(3)

print("First 3 rows of DataFrame:")
print(first_three_rows)
````


# Q6. Select Data using iloc

## Problem Statement
Create a DataFrame with at least 4 rows.

Using `iloc`, print:
- Second row
- First two columns

## Sample Output
```python
Selected row values
```


````python
import pandas as pd

# 1. Create the DataFrame with at least 4 rows
data = {
    "Name": ["Aman", "Rahul", "Neha", "Priya"],
    "Age": [21, 22, 20, 23],
    "City": ["Bhopal", "Indore", "Delhi", "Mumbai"]
}
df = pd.DataFrame(data)

# 2. Using iloc, print the second row and first two columns
print("Selected row values (Second row):")
print(df.iloc[1])

print("\nFirst two columns:")
print(df.iloc[:, :2])
````

## Problem Statement
Create a DataFrame:


Rename column `Marks` to `Score`.


# Q8. Handle Missing Values

## Problem Statement
Create a DataFrame:

```python
Name = ["Aman", None, "Neha"]
Marks = [80, 75, None]
```

Fill missing values using `fillna()`.

## Sample Output
```python
Missing values replaced
```


````python
import pandas as pd
import numpy as np

# 1. Create the DataFrame with missing values
data = {
    "Name": ["Aman", None, "Neha"],
    "Marks": [80, 75, None]
}
df = pd.DataFrame(data)

# 2. Fill missing values using fillna()
filled_df = df.fillna({"Name": "Unknown", "Marks": df["Marks"].mean()})
print("Missing values replaced:")
print(filled_df)
````


# Q9. Replace Values

## Problem Statement
Create a DataFrame:

```python
City = ["BPL", "IND", "BPL"]
```

Replace:
- BPL → Bhopal
- IND → Indore

## Sample Output
```python
Bhopal
Indore
Bhopal
```


````python
import pandas as pd

# 1. Create the DataFrame
data = {
    "City": ["BPL", "IND", "BPL"]
}
df = pd.DataFrame(data)

# 2. Replace values
df["City"] = df["City"].replace({"BPL": "Bhopal", "IND": "Indore"})
print(df["City"].values)
````


# Q10. Remove Duplicate Records

## Problem Statement
Create a DataFrame:

```python
Name = ["Aman","Rahul","Aman","Neha"]
```

Check duplicate rows using `duplicated()` and remove them using `drop_duplicates()`.

## Sample Output
```python
Aman
Rahul
Neha
```


````python
import pandas as pd

# 1. Create the DataFrame with duplicates
data = {
    "Name": ["Aman", "Rahul", "Aman", "Neha"]
}
df = pd.DataFrame(data)

# 2. Check duplicates and remove them
print("Duplicate rows mask:")
print(df.duplicated())

cleaned_df = df.drop_duplicates()
print("\nAfter removing duplicates:")
print(cleaned_df)
````
