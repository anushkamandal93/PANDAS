## 3 Understanding Series & DataFrame
- **Series**: One-dimensional labeled array, like a column in Excel.
- **DataFrame**: Two-dimensional table, similar to an Excel sheet or SQL table.

### Creating a Series from Python Lists or Dictionaries

````python
import pandas as pd
marks = pd.Series([30,40,50,60,70])
marks
````

````python
import pandas as pd
#to know the index
marks = pd.Series([30,40,50,60,70],index=['vihaan','shaurya','aveek', 'kiana','arhaaan'])
marks['aveek']
````

````python
import pandas as pd
#to know the index
marks = pd.Series([30,40,50,60,70])
marks[4]
````

### Creating a DataFrame

````python
#mutliple col. --> dataframe
info = pd.DataFrame(
{
    'name':['vihaan','shaurya','aveek', 'kiana','arhaaan'],
    'age':[16,18,20,19,17],
    'marks':[45,35,50,48,33],
})
info

#selecting a col.
print(info['name'])

#to get specific row we use .loc
print([info.loc[1]])

#select a specific value
print(info.loc[1,'name'])
````

### Creating DataFrame from NumPy Arrays

````python
information=pd.DataFrame([
 ['rahul',52,80],
['devraj',22,85],
['abhishek',60,56] ], columns=['name','age','marks'])

print(information)
print(information.loc[1],'name')

````

````python
#generate a 3x4 matrix with random numbers(1-13)using numpy and name the col. as [A,B,C,D]
import numpy as np
import pandas as pd
arr=np.random.randint(1,100,size=(3,4))

arr_df=pd.DataFrame(arr,columns=['A','B','C','D'])
print(arr_df)
````
